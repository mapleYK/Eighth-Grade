```python
import random
pl_pos = 1
c_pos = 1
def banmen():
    print("."*(pl_pos-1)+"p"+"."*(30-pl_pos)+"GOAL")
    print("."*(c_pos-1)+"c"+"."*(30-c_pos)+"GOAL")

banmen()

while True:
    input("按下enter就會前進")
    pl_pos += random.randint(1,6)
    if pl_pos > 30:
        pl_pos = 30
    banmen()
    if pl_pos == 30:
        print("你獲勝了!")
        break
        
    c_pos += random.randint(1,6)
    if c_pos > 30:
        c_pos = 30
    banmen()
    
    if c_pos == 30:
        print("電腦獲勝了!")
        break
```

```python
#修改成公平一點，不要印那麼多次
import random
pl_pos = 1
c_pos = 1
def banmen():
    print("."*(pl_pos-1)+"p"+"."*(30-pl_pos)+"GOAL")
    print("."*(c_pos-1)+"c"+"."*(30-c_pos)+"GOAL")

banmen()

while True:
    input("按下enter就會前進")
    pl_pos += random.randint(1,6)
    c_pos += random.randint(1,6)
    if pl_pos > 30:
        pl_pos = 30
    if c_pos > 30:
        c_pos = 30  
    banmen()
    if pl_pos == 30 and c_pos !=30:
        print("你獲勝了!")
        break
    elif c_pos == 30 and pl_pos !=30:
        print("電腦獲勝了!")
        break
    elif pl_pos == 30 and c_pos == 30:
        print("平手")    
```
