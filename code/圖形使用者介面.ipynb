```python
import tkinter as tk
root = tk.Tk() 
root.title("視窗")
root.geometry("800x600")
label = tk.Label(root, text = "標籤的字串", font = ("System",24)) #建立標籤零件(物件)
label.place(x=200,y=100)#左上角為(0,0)，'往下往右為y,x之正向
root.mainloop()
```

```python
#配置按鈕
import tkinter as tk
root = tk.Tk() 
root.title("視窗")
root.geometry("800x600")
button = tk.Button(root, text = "按鈕的字串", font = ("微軟正黑體",24)) #建立標籤零件(物件)
button.place(x=200,y=100)#左上角為(0,0)，往下往右為y,x之正向
root.mainloop()
```

```python
#配置按鈕
import tkinter as tk

def click_btn():#宣告btn函數，用來變更字串按鈕
    button["text"] = "點選按鈕了"

root = tk.Tk() 
root.title("視窗")
root.geometry("800x600")
button = tk.Button(root, text = "按鈕的字串", font = ("微軟正黑體",24),command = click_btn) #建立標籤零件(物件)
button.place(x=200,y=100)#左上角為(0,0)，往下往右為y,x之正向
root.mainloop()
```

```python
#colab沒有桌面GUI
import tkinter
root = tkinter.Tk()
root.title("123456")
canvas = tkinter.Canvas(root, width=400, height=600)
canvas.pack()
dog = tkinter.PhotoImage(file="123456.png")
canvas.create_image(200,300,image=dog)
root.mainloop()
```
