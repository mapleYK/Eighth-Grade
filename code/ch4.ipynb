```python
#透過while迴圈找到特定字母的字串
fruits =  ["apple", "banana", "cherry"]
letter = "a"
index = 0
found = False

while not found and index < len(fruits):
    if letter in fruits[index]:
        found = True
    else:
        index += 1
if found:
    print(fruits[index])
else:
    print("沒有找到")
```

```python
star = 9
while star > 0:
    for i in range(1,star):
        print("*", end = "")
    print("*")
    star -= 2
```

```python
space = 0
star = 17
while star > 0:
    for i in range(space):
        print(" ", end = "")
        
    for i in range(1,star):
        print("*", end = "")
    print("*")
    star -= 4
    space += 2
```
