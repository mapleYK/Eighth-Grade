```python
import random
ALP = ["A","B","C","D","E","F","G"]
r = random.choice(ALP)
alp = ""

for i in ALP:
    if i != r:
        alp = alp + i
print("alp")

ans = input("缺少哪個字母?")

if ans == r:
    print("答案正確")
else:
    print("答案錯誤")
```

```python
#加入時間計測
import random
import datetime

ALP = ["A","B","C","D","E","F","G"]
r = random.choice(ALP)
alp = ""

for i in ALP:
    if i != r:
        alp = alp + i
print(alp)

st = datetime.datetime.now()

ans = input("缺少哪個字母?")

if ans == r:
    print("答案正確")
    et = datetime.datetime.now()
    print((et-st).seconds)
else:
    print("答案錯誤")
```
