```python
#運算子
x = 10
y = 2
print(x/y)
print(x//y)

z = 4
print(x/z)
print(x//z)

print(x%z)
```

```python
#判斷是否為偶數
number = input("請輸入數字:")
number = int(number)

if number % 2 == 0:
    print(number, "是偶數")
else:
    print(number, "是奇數")
```

```python
#判斷輸入的字串是否為空
string = ""
if len(string) == 0 :
    print("字串為空")
```

```python
#判斷一個字串是否包含特定字元
string = "hello"
if "e" in string:
    print("字串包含e")
```

```python
#判斷一個數字是否在範圍內
num = 5
if num in range(1,10):
    print("數字在範圍內")
```

```python
#配合邏輯運算子
a = True
b = False

if a and b:
    print("兩個都是真")
else:
    print("至少有一個不為真")
```

```python
#elif
x = input("請輸入一個數:")
x = int(x)

if x < 5:
    print("x小於5")
elif x < 10:
    print("x介於5~9之間")
elif x < 15:
    print("x介於10~14之間")
else:
    print("x大於14")
```

```python
x = int(input("請輸入數字x:"))
y = int(input("請輸入數字y:"))
z = int(input("請輸入數字z:"))

if x > y:
    print("x > y")
    if x > z:
        print("x > z")
    else:
        print("x <= z")
else:
    print("x <= y")
```
